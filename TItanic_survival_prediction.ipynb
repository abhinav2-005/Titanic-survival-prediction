```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset("titanic")
```

```python
df
```

```python
df.head()
```

```python
df.tail()
```

```python
df.columns
```

```python
df.isna().sum()
```

```python
df.drop(["deck","embark_town","alive","class","who","adult_male"],axis = 1,inplace = True)
```

```python
df.head()
```

```python
df['age'].fillna(df['age'].mean(),inplace = True)
```

```python
df.head()
```

```python
df.dropna(subset=["embarked"],inplace = True)
```

```python
df.head()
```

```python
df.isna().sum()
```

```python
df['alone'] = df['alone'].astype(int)
```

```python
df.head()
```

```python
df['sex'] = df['sex'].map({"male":0,"female":1})
```

```python
df.head()
```

```python
df = pd.get_dummies(df,columns = ["embarked"])
```

```python
df['embarked_C'] = df['embarked_C'].astype(int)
df['embarked_Q'] = df['embarked_Q'].astype(int)
df['embarked_S'] = df['embarked_S'].astype(int)
```

```python
df.head()
```

```python
x = df.drop(['survived'],axis = 1)
y = df['survived']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test, y_train, y_test = train_test_split(x, y, test_size=0.2,random_state=42)
```

Logistic regression

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()
model.fit(x_train,y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

accuracy_score(y_test,y_pred)
```

```python
confusion_matrix(y_test,y_pred)
```

```python
print(classification_report(y_test,y_pred))
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

```

```python
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.fit_transform(x_test)
```

KNN

```python
from sklearn.neighbors import KNeighborsClassifier
knn_model = KNeighborsClassifier(n_neighbors = 5)
knn_model.fit(x_train_scaled,y_train)
```

```python
y_pred_knn = knn_model.predict(x_test_scaled)
```

```python
accuracy_score(y_test,y_pred_knn)
```

```python
print(classification_report(y_test,y_pred_knn))
```

navie bayes algorithm

```python
from sklearn.naive_bayes import GaussianNB
```

```python
navie_model = GaussianNB()

navie_model.fit(x_train,y_train)

```

```python
y_pred_navie = navie_model.predict(x_test)
```

```python
accuracy_score(y_test,y_pred_navie)
```

```python
print(classification_report(y_test,y_pred_navie))
```

decision tree

```python
from sklearn.tree import DecisionTreeClassifier

model_desicion_tree = DecisionTreeClassifier()
```

```python
model_desicion_tree.fit(x_train_scaled,y_train)
```

```python
y_pred_dt = model_desicion_tree.predict(x_test_scaled)
```

```python
accuracy_score(y_test,y_pred_dt)
```

```python
print(classification_report(y_test,y_pred_dt))
```

Support vector machine (SVM)

```python
from sklearn.svm import SVC
model_svc = SVC(kernel = 'rbf')
model_svc.fit(x_train_scaled,y_train)
```

```python
y_pred_svc = model_svc.predict(x_test_scaled)
```

```python
accuracy_score(y_test,y_pred_svc)
```

```python
print(classification_report(y_test,y_pred_svc))
```

```python
confusion_matrix(y_test,y_pred_svc)
```

cross valiadition

```python
from sklearn.model_selection import cross_val_score

x_train = df.drop('survived',axis = 1)
y = df['survived']

x_train_scaled = scaler.fit_transform(x_train)

scores = cross_val_score(model_svc, x_train_scaled, y, cv = 5, scoring = 'accuracy')
```
